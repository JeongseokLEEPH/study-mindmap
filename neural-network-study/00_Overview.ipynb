# 🧠 딥러닝 핵심 알고리즘 마스터 (직접 실행하며 배우기)

이 저장소는 **인공신경망/딥러닝의 필수 알고리즘**들을 **직접 코드를 실행해보면서** 배우는 학습용 노트북 모음입니다.
모든 코드는 외부 데이터 다운로드 없이 ** 코드만으로 합성 데이터를 만들어** 실행되도록설계했습니다.

## 📚 학습 로드맵 (권장 순서)

| 순서 | 파일 | 알고리즘 | 핵심 아이디어 | 직접 해보는 예시 |
|---|---|---|---|---|
| 1 | `01_Perceptron_ADALINE.ipynb` | 단층 퍼셉트론, ADALINE | 가중치 갱신 규칙, 결정 경계 | AND 게이트, XOR 한계, 2D 분류 |
| 2 | `02_MLP.ipynb` | MLP (다층 퍼셉트론) | 역전파(Backpropagation) | XOR 해결, moons 데이터 분류 |
| 3 | `03_Boltzmann_Machine.ipynb` | RBM (제한된 볼츠만 머신) | 에너지 기반 모델, CD-1, 깁스 샘플링 | 3x3 막대 패턴 생성 |
| 4 | `04_CNN.ipynb` | CNN | 합성곱, 풀링, 가중치 공유 | 원/사각형/삼각형 이미지 분류 |
| 5 | `05_RNN.ipynb` | RNN / LSTM / GRU | 순환 구조, 은닉 상태 | 사인파 다음 값 예측 비교 |
| 6 | `06_GNN.ipynb` | GNN (GCN) | 그래프 메시지 패싱 |커뮤니티 노드 분류 |
| 7 | `07_GAN.ipynb` | GAN | 생성자 vs 판별자 minmax 게임 | 2D 8-가우시안 분포 생성 |
| 8 | `08_Transformer.ipynb` | Transformer | 셀프 어텐션 (Q/K/V) | 정렬된 수열 판별 + 어텐션 시각화 |
| 9 | `09_AutoRegressive.ipynb` | AR (자기회귀), MADE | p(x)=∏p(xᵢ\|x<ᵢ) 분해 | AR(2) 시계열 예측, 3D 분포 모델링 |
